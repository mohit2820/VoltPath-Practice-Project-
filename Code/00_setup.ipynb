## Dynamic Catalog

```python
# Catalog parameter
dbutils.widgets.text("CATALOG", "vp_vk_mkc20040528")
CATALOG = dbutils.widgets.get("CATALOG")
```

## Create catalog

```python
%sql
-- Create catalog
CREATE CATALOG IF NOT EXISTS IDENTIFIER(:CATALOG)
COMMENT 'VoltPath EV charging data platform';
```

## Create schemas

```python
%sql
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:CATALOG || '.landing')
COMMENT 'Landing data';

CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:CATALOG || '.bronze')
COMMENT 'Bronze layer';

CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:CATALOG || '.silver')
COMMENT 'Silver layer';

CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:CATALOG || '.gold')
COMMENT 'Gold layer';

CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:CATALOG || '.ops')
COMMENT 'Pipeline operations';
```

## Create managed volumes

```python
%sql
CREATE VOLUME IF NOT EXISTS IDENTIFIER(:CATALOG || '.landing.raw')
COMMENT 'Raw source files';

CREATE VOLUME IF NOT EXISTS IDENTIFIER(:CATALOG || '.ops.pipeline_state')
COMMENT 'Pipeline checkpoint and state files';
```

## Create directories

```python
BASE_PATH = f"/Volumes/{CATALOG}/landing/raw"

for folder in ["stations", "tariffs", "chargers", "members", "sessions"]:
    dbutils.fs.mkdirs(f"{BASE_PATH}/{folder}")
```
